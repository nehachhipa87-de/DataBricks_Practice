```python
%sql
COPY INTO neha_catalog.bronze_schema.rawMonthlySales
FROM (
    SELECT sale_id, product, category, quantity, price, sale_date, region,current_timestamp() AS ingest_ts
    FROM '/Volumes/neha_catalog/bronze_schema/landing_files/src_files/monthly_sales/'
    )
FILEFORMAT = CSV
FORMAT_OPTIONS('header'='true')
COPY_OPTIONS('mergeSchema'='true')
```
